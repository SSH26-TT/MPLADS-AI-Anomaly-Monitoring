```python
# ============================================================
# 03 - MODEL EXPORT & FINAL VERIFICATION
# ============================================================

import pandas as pd
import numpy as np
import joblib
import json

print("Financial Anomaly Detection")
print("Model Export & Final Verification")
print("=" * 50)
```

```python
# Load exact validated model

model = joblib.load(
    "isolation_forest_model.pkl"
)

# Load exact feature list

with open("model_features.json", "r") as f:
    model_features = json.load(f)

# Load risk thresholds

with open("risk_thresholds.json", "r") as f:
    risk_thresholds = json.load(f)

print("Model loaded successfully.")

print("\nFeatures:")
for i, feature in enumerate(model_features, 1):
    print(f"{i}. {feature}")

print("\nRisk thresholds:")
for name, value in risk_thresholds.items():
    print(f"{name}: {value}")
```

```python
test_df = pd.read_csv(
    "financial_anomaly_test_final.csv"
)

print("Test dataset loaded.")
print("Shape:", test_df.shape)

display(test_df.head())
```

```python
X_test = test_df[model_features].copy()

print("X_test shape:", X_test.shape)
print("Missing values:", X_test.isna().sum().sum())
```

```python
# ============================================================
# EXPORTED MODEL VERIFICATION
# ============================================================

raw_scores = model.score_samples(X_test)

anomaly_scores = -raw_scores

predictions = model.predict(X_test)

anomaly_flags = (
    predictions == -1
).astype(int)

print("=" * 60)
print("EXPORTED MODEL VERIFICATION")
print("=" * 60)

print("\nTest works:", len(X_test))

print("Anomalous works:", anomaly_flags.sum())

print(
    "Anomaly percentage:",
    round(anomaly_flags.mean() * 100, 4),
)

print(
    "\nScore range:",
    round(anomaly_scores.min(), 6),
    "to",
    round(anomaly_scores.max(), 6)
)
```
